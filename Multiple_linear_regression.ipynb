```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Dataset: Experience and Hours Studied
X = np.array([
    [1, 2],
    [2, 3],                                  #ap24110010586
    [3, 4],
    [4, 5],
    [5, 6]
])

# Output: Salary
Y = np.array([20, 30, 40, 50, 60])

# Create and train the model
model = LinearRegression()
model.fit(X, Y)

# Prediction
Y_pred = model.predict(X)

# Display results
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
print("Predicted values:", Y_pred)

# Predict for new data
new_data = np.array([[6, 7]])
print("Prediction:", model.predict(new_data)[0])

```

Multiple Linear Regression

Predicts a dependent variable using two or more independent variables.

Formula: (Y = b_0 + b_1X_1 + b_2X_2)

Example: Predicting marks based on study hours and attendance.
