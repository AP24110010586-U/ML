##PCA (Principal Component Analysis) in Python


```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Create a dataset
data = {
    'A': [2, 4, 6, 8, 10],
    'B': [1, 3, 5, 7, 9],
    'C': [5, 10, 15, 20, 25]
}                                          #AP24110010586

df = pd.DataFrame(data)

# Step 1: Standardize the data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

# Step 2: Apply PCA
pca = PCA(n_components=2)
pca_data = pca.fit_transform(scaled_data)

# Step 3: Display results
print("Original Data:")
print(df)

print("\nPCA Transformed Data:")
print(pca_data)

print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

print("\nTotal Variance Preserved:")
print(sum(pca.explained_variance_ratio_))
```

Import Pandas, StandardScaler, and PCA from Scikit-learn. Create a dataset containing multiple numerical features and load it into a Pandas DataFrame. Standardize the dataset using StandardScaler() to normalize the scale of all variables. Next, initialize PCA with n_components=2 to compress the dimensionality down to two principal components, and apply fit_transform() to compute the transformation. Display both the raw input dataset and the transformed output dataset. check explained_variance_ratio_ to evaluate the proportion of dataset variance captured by each principal component. PCA effectively compresses feature space while retaining as much original information as possible.
